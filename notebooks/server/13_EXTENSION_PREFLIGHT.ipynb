# 13 — Paper-extension preflight
Khóa protocol và kiểm tra đầu vào trước khi chạy các thí nghiệm phục vụ bài báo. Notebook không thay đổi 120 run chính.

```python
from pathlib import Path
import hashlib, json, os, subprocess, sys
ROOT = Path.cwd()
if not (ROOT / 'research/server').is_dir():
    raise RuntimeError('Hãy mở Jupyter tại thư mục gốc của repo')
PYTHON = os.environ.get('NIDS_PYTHON', str(ROOT / '.venv-server/bin/python'))
def sha256(path):
    return hashlib.sha256(Path(path).read_bytes()).hexdigest()
def stage(name):
    subprocess.run(['bash', 'research/server/run_next_experiments.sh', name], cwd=ROOT, check=True)
```

```python
protocol = ROOT / 'research/PROTOCOL_PAPER_EXTENSION_VI.md'
required = [protocol, ROOT / 'data/full_splits', ROOT / 'research/artifacts/full_runs_5seed/runs.csv']
missing = [str(path) for path in required if not path.exists()]
if missing:
    raise FileNotFoundError(missing)
baseline = ROOT / 'research/artifacts/full_runs_5seed/runs.csv'
import pandas as pd
runs = pd.read_csv(baseline)
assert len(runs) == 120 and runs.seed.nunique() == 5
print({'protocol_sha256': sha256(protocol), 'verified_base_runs': len(runs), 'seeds': sorted(runs.seed.unique().tolist())})
```
